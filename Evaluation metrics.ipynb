# Performance Evaluation

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, root_mean_squared_error, mean_absolute_percentage_error
import numpy as np
import pandas as pd
```

## Reading serieses

```python
truth = pd.read_csv('truth.csv')
model = pd.read_csv('new_attention.csv')
cnn = pd.read_csv('cnn_predictions_300.csv')
cnnlstm = pd.read_csv('cnnlstm_predictions_300.csv')
cnngru = pd.read_csv('cnngru_predictions_300.csv')
```

## denormalization

```python
truth['Accelerator'] = truth['Accelerator'] * 100
model['Accelerator'] = model['Accelerator'] * 100
cnn['Accelerator'] = cnn['Accelerator'] * 100
cnnlstm['Accelerator'] = cnnlstm['Accelerator'] * 100
cnngru['Accelerator'] = cnngru[['Accelerator']] * 100

truth['Steering Wheel'] = (truth['Steering Wheel'] * 1200 ) - 600
model['Steering Wheel'] = (model['Steering Wheel'] * 1200 ) - 600
cnn['Steering Wheel'] = (cnn['Steering Wheel'] * 1200 ) - 600
cnnlstm['Steering Wheel'] = (cnnlstm['Steering Wheel'] * 1200) - 600
cnngru['Steering Wheel'] = (cnngru['Steering Wheel'] * 1200) - 600

# brake is already okay
```

## $R^2$ coefficient of determination

```python
r2 = r2_score(truth['Accelerator'][9:], model['Accelerator'])
print(f'model r2 for acclerator: {r2}')
r2 = r2_score(truth['Steering Wheel'][9:], model['Steering Wheel'])
print(f'model r2 for steering wheel: {r2}')
r2 = r2_score(truth['Brake'][9:], model['Brake'])
print(f'model r2 for brake pedal: {r2}')

r2 = r2_score(truth['Accelerator'], cnn['Accelerator'])
print(f'cnn r2 for acclerator: {r2}')
r2 = r2_score(truth['Steering Wheel'], cnn['Steering Wheel'])
print(f'cnn r2 for steering wheel: {r2}')
r2 = r2_score(truth['Brake'], cnn['Brake'])
print(f'cnn r2 for brake pedal: {r2}')

r2 = r2_score(truth['Accelerator'][9:], cnnlstm['Accelerator'])
print(f'cnnlstm r2 for acclerator: {r2}')
r2 = r2_score(truth['Steering Wheel'][9:], cnnlstm['Steering Wheel'])
print(f'cnnlstm r2 for steering wheel: {r2}')
r2 = r2_score(truth['Brake'][9:], cnnlstm['Brake'])
print(f'cnnlstm r2 for brake pedal: {r2}')

r2 = r2_score(truth['Accelerator'][9:], cnngru['Accelerator'])
print(f'cnngru r2 for acclerator: {r2}')
r2 = r2_score(truth['Steering Wheel'][9:], cnngru['Steering Wheel'])
print(f'cnngru r2 for steering wheel: {r2}')
r2 = r2_score(truth['Brake'][9:], cnngru['Brake'])
print(f'cnngru r2 for brake pedal: {r2}')
```

```python
mse = mean_squared_error(truth['Accelerator'][9:], model['Accelerator'])
print(f'model mse for acclerator: {mse}')
mse = mean_squared_error(truth['Steering Wheel'][9:], model['Steering Wheel'])
print(f'model mse for steering wheel: {mse}')
mse = mean_squared_error(truth['Brake'][9:], model['Brake'])
print(f'model mse for brake pedal: {mse}')

mse = mean_squared_error(truth['Accelerator'], cnn['Accelerator'])
print(f'cnn mse for acclerator: {mse}')
mse = mean_squared_error(truth['Steering Wheel'], cnn['Steering Wheel'])
print(f'cnn mse for steering wheel: {mse}')
mse = mean_squared_error(truth['Brake'], cnn['Brake'])
print(f'cnn mse for brake pedal: {mse}')

mse = mean_squared_error(truth['Accelerator'][9:], cnnlstm['Accelerator'])
print(f'cnnlstm mse for acclerator: {mse}')
mse = mean_squared_error(truth['Steering Wheel'][9:], cnnlstm['Steering Wheel'])
print(f'cnnlstm mse for steering wheel: {mse}')
mse = mean_squared_error(truth['Brake'][9:], cnnlstm['Brake'])
print(f'cnnlstm mse for brake pedal: {mse}')

mse = mean_squared_error(truth['Accelerator'][9:], cnngru['Accelerator'])
print(f'cnngru mse for acclerator: {mse}')
mse = mean_squared_error(truth['Steering Wheel'][9:], cnngru['Steering Wheel'])
print(f'cnngru mse for steering wheel: {mse}')
mse = mean_squared_error(truth['Brake'][9:], cnngru['Brake'])
print(f'cnngru mse for brake pedal: {mse}')
```
