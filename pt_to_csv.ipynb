```python
import torch
import pandas as pd
```

```python
tensor_file = "./swin_predictions.pt"
csv_file_name = "new_attention.csv"
```

```python
record = torch.load(tensor_file, weights_only=True).tolist()

record_df = pd.DataFrame(csv_file_name, columns=["Accelerator", "Steering Wheel", "Brake"])
record_df.to_csv("./new_attention.csv", index=False)
```

```python
tensor_file = "predictions_cnngru_300.pt"
csv_file_name = "cnngru_predictions_300.csv"
```
